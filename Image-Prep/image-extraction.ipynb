```python
import numpy as np
import cv2
import os
import pickle
import pandas as pd
from tqdm import tqdm

from dataset.PIE_API import PIE
from tools.video_utils import VideoLoader
from tools.visualisation import draw_bbox
from tools.box_utils import jitter_bbox, squarify_bbox

# CONFIGURATION
DATASET_DIR = r"..\PIE-Files"
CLIPS_DIR   = os.path.join(DATASET_DIR, "PIE_clips")
DB_PATH     = os.path.join(DATASET_DIR, "data_cache", "pie_database.pkl")
TARGET_DIM  = (224, 224)
SAVE_DIR    = "../Bounding-Box"
SAVE_TOGGLE = True ### Change Toggle to Save
os.makedirs(SAVE_DIR, exist_ok=True)

ATTR_CSVS = {
    "FN": "../Data-Prep/annot-data/PIE_annot_attrb_val_fn.csv",
    "FP": "../Data-Prep/annot-data/PIE_annot_attrb_val_fp.csv",
    "TP": "../Data-Prep/annot-data/PIE_annot_attrb_val_tp.csv",
    "TN": "../Data-Prep/annot-data/PIE_annot_attrb_val_tn.csv"
}


def show_local_pedestrian_clip(subset="FP", ped_id="3_2_295", frames_to_show=32, save_output=SAVE_TOGGLE):
    print(f"🎬 Generating local clip for {subset} → {ped_id}")

    if not os.path.isfile(DB_PATH):
        print("⚠️ PIE database not found — generating...")
        dataset = PIE(data_path=DATASET_DIR)
        dataset.generate_database()
    data = pickle.load(open(DB_PATH, "rb"))

    attr_path = ATTR_CSVS.get(subset)
    attr_df = pd.read_csv(attr_path)
    row = attr_df[attr_df["id"] == ped_id]
    critical_point = int(row["critical_point"].iloc[0])

    set_id, vid_id, ped_num = ped_id.split("_")
    set_name = f"set0{set_id[-1]}" if not set_id.startswith("set") else set_id
    video_name = f"video_{int(vid_id):04d}"
    video_path = os.path.join(CLIPS_DIR, set_name, f"{video_name}.mp4")
    video = VideoLoader(video_path)
    annots = data[set_name][video_name]["ped_annotations"].get(ped_id)
    frames = annots["frames"]

    crit_idx = frames.index(critical_point)
    start_idx = max(0, crit_idx - (frames_to_show - 1))
    seq_frames = frames[start_idx:crit_idx + 1]

    if save_output:
        ped_outdir = os.path.join(SAVE_DIR, f"{subset}/{ped_id}")
        os.makedirs(ped_outdir, exist_ok=True)
        out_path = os.path.join(ped_outdir, f"{subset}_{ped_id}_local.mp4")
        fourcc = cv2.VideoWriter_fourcc(*"mp4v")
        video_writer = cv2.VideoWriter(out_path, fourcc, 10, TARGET_DIM)
        print(f"💾 Saving cropped video to {out_path}")

    # Step 4: Main frame loop
    for frame_idx in seq_frames:
        frame = video[frame_idx]
        t = annots["frames"].index(frame_idx)
        bbox = list(map(int, annots["bbox"][t][0:4]))

        # Expand + squarify for crop
        jitter_box = jitter_bbox(video.size, [bbox.copy()], mode="enlarge", ratio=1.0)[0]
        local_box = squarify_bbox(jitter_box, 1.0, frame.shape[1])
        local_box = tuple(map(int, local_box[0:4]))

        local_img = frame[local_box[1]:local_box[3], local_box[0]:local_box[2]]
        if local_img.size == 0:
            continue

        local_resized = cv2.resize(local_img, TARGET_DIM)

        pad_ratio = 0.0  # 10% padding relative to bbox size
        x1, y1, x2, y2 = bbox
        w, h = x2 - x1, y2 - y1
        pad_x, pad_y = w * pad_ratio, h * pad_ratio
        padded_box = [x1 - pad_x, y1 - pad_y, x2 + pad_x, y2 + pad_y]

        # Map to resized crop coordinates
        scale_x = TARGET_DIM[0] / (local_box[2] - local_box[0])
        scale_y = TARGET_DIM[1] / (local_box[3] - local_box[1])
        mapped_box = [
            (padded_box[0] - local_box[0]) * scale_x,
            (padded_box[1] - local_box[1]) * scale_y,
            (padded_box[2] - local_box[0]) * scale_x,
            (padded_box[3] - local_box[1]) * scale_y,
        ]

        # Clamp to image boundaries
        mapped_box = [max(0, mapped_box[0]), max(0, mapped_box[1]),
                    min(TARGET_DIM[0], mapped_box[2]), min(TARGET_DIM[1], mapped_box[3])]

        # Draw per-frame adaptive box
        annotated = draw_bbox(local_resized.copy(), mapped_box,
                            caption=f"{ped_id}", thickness=2, color=(255, 0, 0))
        cv2.imshow("Local Pedestrian View", annotated)

        if save_output:
            video_writer.write(annotated)
            cv2.imwrite(os.path.join(ped_outdir, f"frame_{frame_idx}.jpg"), annotated)

        key = cv2.waitKey(60)
        if key == ord("q"):
            break

    cv2.destroyAllWindows()
    if save_output:
        video_writer.release()
        print(f"✅ Saved to: {out_path}")

if __name__ == "__main__":
    # subsets = ['FN', 'FP', 'TP', 'TN']
    subsets = ['TP','TN']
    for subset in subsets:
        ped_set = []
        if subset == 'FN': ped_set = fn_occ_top_30
        elif subset == 'FP': ped_set = fp_occ_top_30
        elif subset == 'TP': ped_set = tp_occ_combined
        elif subset == 'TN': ped_set = tn_occ_combined

        for ped in ped_set:
            show_local_pedestrian_clip(subset=subset, ped_id=ped, frames_to_show=32, save_output=SAVE_TOGGLE)
```

```python
import os
import sys
import cv2
import pickle
import numpy as np
import pandas as pd
from tqdm import tqdm
import importlib
YOLO_ROOT = os.path.join(os.getcwd(), "feature", "yolov7pose")

from dataset.PIE_API import PIE
from tools.video_utils import VideoLoader
from tools.pose_utils import draw_body_keypoints
from tools.box_utils import jitter_bbox, squarify_bbox, calculate_iou
from tools.visualisation import draw_bbox

sys.modules["utils"] = importlib.import_module("feature.yolov7pose.utils")
sys.modules["models"] = importlib.import_module("feature.yolov7pose.models")

from feature.yolov7pose._interface import PoseEstimator

DATASET_DIR   = r"..\PIE-Files"
CLIPS_DIR     = os.path.join(DATASET_DIR, "PIE_clips")
DB_PATH       = os.path.join(DATASET_DIR, "data_cache", "pie_database.pkl")
OUTPUT_DIR    = r"../Body-Pose"
TARGET_DIM    = (256,256)
SAVE_TOGGLE = True

os.makedirs(OUTPUT_DIR, exist_ok=True)

ATTR_CSVS = {
    "FN": "../Data-Prep/annot-data/PIE_annot_attrb_val_fn.csv",
    "FP": "../Data-Prep/annot-data/PIE_annot_attrb_val_fp.csv",
    "TP": "../Data-Prep/annot-data/PIE_annot_attrb_val_tp.csv",
    "TN": "../Data-Prep/annot-data/PIE_annot_attrb_val_tn.csv"
}

Pose = PoseEstimator(
    weights=os.path.join(YOLO_ROOT, "weights_pose.pt"),
    device="cpu"
)

dataset = PIE(data_path=DATASET_DIR)
if not os.path.isfile(DB_PATH):
    dataset.generate_database()
data = pickle.load(open(DB_PATH, "rb"))

def extract_pose_frames(subset, ped_id, frames_to_show=32, save_output=True):
    print(f"Extracting pose → {subset}/{ped_id}")
    attr_path = ATTR_CSVS.get(subset)
    attr_df = pd.read_csv(attr_path)
    row = attr_df[attr_df["id"] == ped_id]

    if row.empty:
        print("Pedestrian not found")
        return

    critical_point = int(row["critical_point"].iloc[0])
    set_id, vid_id, ped_num = ped_id.split("_")
    set_name = f"set0{set_id[-1]}" if not set_id.startswith("set") else set_id
    video_name = f"video_{int(vid_id):04d}"
    video_path = os.path.join(CLIPS_DIR, set_name, f"{video_name}.mp4")
    video = VideoLoader(video_path)

    annots = data[set_name][video_name]["ped_annotations"].get(ped_id)
    frames = annots["frames"]
    crit_idx = frames.index(critical_point)
    start_idx = max(0, crit_idx - (frames_to_show - 1))
    seq_frames = frames[start_idx:crit_idx + 1]

    if save_output:
        ped_outdir = os.path.join(OUTPUT_DIR, subset, ped_id)
        os.makedirs(ped_outdir, exist_ok=True)

    for frame_idx in seq_frames:
        frame = video[frame_idx]
        t = annots["frames"].index(frame_idx)
        bbox = list(map(int, annots["bbox"][t][0:4]))
        jitter_box = jitter_bbox(video.size, [bbox.copy()], mode="enlarge", ratio=1.0)[0]
        local_box = squarify_bbox(jitter_box, 1.0, frame.shape[1])
        local_box = tuple(map(int, local_box[0:4]))
        local_img = frame[
            local_box[1]:local_box[3],
            local_box[0]:local_box[2]
        ]

        if local_img.size == 0:
            continue
        safe_resized = cv2.resize(local_img, TARGET_DIM)
        scale_x = TARGET_DIM[0] / (local_box[2] - local_box[0])
        scale_y = TARGET_DIM[1] / (local_box[3] - local_box[1])
        mapped_box = [
            (bbox[0] - local_box[0]) * scale_x,
            (bbox[1] - local_box[1]) * scale_y,
            (bbox[2] - local_box[0]) * scale_x,
            (bbox[3] - local_box[1]) * scale_y
        ]
        mapped_box = [
            max(0, mapped_box[0]),
            max(0, mapped_box[1]),
            min(TARGET_DIM[0], mapped_box[2]),
            min(TARGET_DIM[1], mapped_box[3])
        ]

        _, output_data = Pose.run(safe_resized, draw_pose=False)

        ped_keypoints = None
        maxIOU = 0

        for pose in output_data:
            for det_index, (*xyxy, conf, cls) in enumerate(pose[:, :6]):
                pose_box = [
                    int(xyxy[0]),
                    int(xyxy[1]),
                    int(xyxy[2]),
                    int(xyxy[3])
                ]
                kpts = pose[det_index, 6:].cpu().numpy()
                iou = calculate_iou(pose_box, mapped_box)
                if iou > maxIOU:
                    maxIOU = iou
                    ped_keypoints = kpts

        annotated = safe_resized.copy()

        # DRAW SKELETON
        if ped_keypoints is not None and maxIOU > 0.3:
            annotated = draw_body_keypoints(annotated, ped_keypoints)

        # DRAW BBOX OVERLAY
        annotated = draw_bbox(
            annotated,
            mapped_box,
            caption=ped_id,
            thickness=2,
            color=(255,0,0)
        )

        cv2.imshow("Pose + BBox", annotated)

        if save_output:
            cv2.imwrite(
                os.path.join(
                    ped_outdir,
                    f"frame_{frame_idx}_pose.jpg"
                ),
                annotated
            )
        if cv2.waitKey(40) == ord("q"):
            break

    cv2.destroyAllWindows()
    print(f"Finished {subset}/{ped_id}")


# MAIN LOOP
if __name__ == "__main__":
    subsets = ["FN","FP","TP","TN"]
    for subset in subsets:
        if subset == "FN": ped_set = fn_occ_top_30
        elif subset == "FP": ped_set = fp_occ_top_30
        elif subset == "TP": ped_set = tp_occ_combined
        elif subset == "TN": ped_set = tn_occ_combined
        else: ped_set = []

        for ped in ped_set:
            extract_pose_frames(
                subset=subset,
                ped_id=ped,
                frames_to_show=32,
                save_output=SAVE_TOGGLE)
```

#### Optical Flow (Local)

```python
import os
import cv2
import pickle
import numpy as np
import pandas as pd
from tqdm import tqdm

import torch
import torchvision.transforms as T
from torchvision.models.optical_flow import raft_large
from torchvision.utils import flow_to_image

from dataset.PIE_API import PIE
from tools.video_utils import VideoLoader
from tools.visualisation import draw_bbox
from tools.box_utils import jitter_bbox, squarify_bbox

# === CONFIGURATION ===
DATASET_DIR = r"..\PIE-Files"
CLIPS_DIR   = os.path.join(DATASET_DIR, "PIE_clips")
DB_PATH     = os.path.join(DATASET_DIR, "data_cache", "pie_database.pkl")

SAVE_DIR    = "../Optical-Flow"
os.makedirs(SAVE_DIR, exist_ok=True)

SAVE_OUTPUT = False

ATTR_CSVS = {
    "FN": "../Data-Prep/annot-data/PIE_annot_attrb_val_fn.csv",
    "FP": "../Data-Prep/annot-data/PIE_annot_attrb_val_fp.csv",
    "TP": "../Data-Prep/annot-data/PIE_annot_attrb_val_tp.csv",
    "TN": "../Data-Prep/annot-data/PIE_annot_attrb_val_tn.csv"
}

# === RAFT MODEL ===
device     = "cuda" if torch.cuda.is_available() else "cpu"
raft_model = raft_large(pretrained=True, progress=False).to(device).eval()

def preprocess_frame(frame):
    transforms = T.Compose([
        T.ToTensor(),
        T.ConvertImageDtype(torch.float32),
        T.Normalize(mean=0.5, std=0.5),
        T.Resize(size=(520, 960)),
    ])
    return transforms(frame).to(device)

def compute_raft_optical_flow(prev_frame, next_frame):
    with torch.no_grad():
        flow     = raft_model(prev_frame.unsqueeze(0), next_frame.unsqueeze(0))[-1]
        flow_img = flow_to_image(flow[0]).byte().permute(1, 2, 0).cpu().numpy()
    return flow_img

# === PIE DATABASE ===
if not os.path.isfile(DB_PATH):
    print("Generating PIE database...")
    dataset = PIE(data_path=DATASET_DIR)
    dataset.generate_database()

data = pickle.load(open(DB_PATH, "rb"))


def extract_flow_frames_local(subset="FP", ped_id="3_2_295", frames_to_show=32):

    print(f"Processing local optical flow {subset} → {ped_id}")

    attr_df = pd.read_csv(ATTR_CSVS[subset])
    row     = attr_df[attr_df["id"] == ped_id]

    if len(row) == 0:
        print("Pedestrian not found in CSV")
        return

    critical_point = int(row["critical_point"].iloc[0])

    set_id, vid_id, ped_num = ped_id.split("_")
    set_name   = f"set0{set_id[-1]}" if not set_id.startswith("set") else set_id
    video_name = f"video_{int(vid_id):04d}"

    video_path = os.path.join(CLIPS_DIR, set_name, f"{video_name}.mp4")
    video      = VideoLoader(video_path)

    annots = data[set_name][video_name]["ped_annotations"].get(ped_id)
    frames = annots["frames"]

    crit_idx   = frames.index(critical_point)
    start_idx  = max(0, crit_idx - (frames_to_show - 1))
    seq_frames = frames[start_idx:crit_idx + 1]

    ped_outdir = os.path.join(SAVE_DIR, f"{subset}/{ped_id}_flow_local")
    if SAVE_OUTPUT:
        os.makedirs(ped_outdir, exist_ok=True)

    TARGET_DIM = (224, 224)
    prev_frame = None

    for frame_idx in seq_frames:
        frame = video[frame_idx]
        t     = annots["frames"].index(frame_idx)
        bbox  = list(map(int, annots["bbox"][t][0:4]))

        frame_h, frame_w = frame.shape[:2]
        x1, y1, x2, y2  = bbox

        # === Local context crop — jitter + squarify matching model input pipeline ===
        jitter_box = jitter_bbox(video.size, [bbox.copy()], mode="enlarge", ratio=1.0)[0]
        local_box  = squarify_bbox(jitter_box, 1.0, frame_w)
        local_box  = tuple(map(int, local_box[0:4]))

        local_img = frame[
            local_box[1]:local_box[3],
            local_box[0]:local_box[2]
        ]

        if local_img.size == 0:
            prev_frame = frame
            continue

        # === Compute RAFT on full frames — preserves scene-level motion context ===
        ref_frame = prev_frame if prev_frame is not None else frame
        flow_full = compute_raft_optical_flow(
            preprocess_frame(ref_frame),
            preprocess_frame(frame)
        )

        # Resize flow output to match full frame dimensions
        flow_full_resized = cv2.resize(
            flow_full,
            (frame_w, frame_h),
            interpolation=cv2.INTER_LINEAR
        )

        # === Crop flow to local box then resize to 224×224 ===
        flow_local = flow_full_resized[
            local_box[1]:local_box[3],
            local_box[0]:local_box[2]
        ]
        flow_local_resized = cv2.resize(
            flow_local,
            TARGET_DIM,
            interpolation=cv2.INTER_LINEAR
        )

        # === Remap bbox into local crop coordinate space ===
        scale_x = TARGET_DIM[0] / (local_box[2] - local_box[0])
        scale_y = TARGET_DIM[1] / (local_box[3] - local_box[1])

        mapped_bbox = [
            max(0,              int((x1 - local_box[0]) * scale_x)),
            max(0,              int((y1 - local_box[1]) * scale_y)),
            min(TARGET_DIM[0],  int((x2 - local_box[0]) * scale_x)),
            min(TARGET_DIM[1],  int((y2 - local_box[1]) * scale_y))
        ]

        annotated = draw_bbox(
            flow_local_resized,
            mapped_bbox,
            caption=ped_id,
            thickness=2,
            color=(255, 0, 0)
        )

        cv2.imshow("Optical Flow Local + BBox", annotated)

        if SAVE_OUTPUT:
            out_path = os.path.join(
                ped_outdir,
                f"frame_{frame_idx}_flow_bbox_local.jpg"
            )
            cv2.imwrite(out_path, annotated)

        prev_frame = frame

        key = cv2.waitKey(1)
        if key == ord("q"):
            break

    cv2.destroyAllWindows()
    print(f"Finished local flow {subset}/{ped_id}")


if __name__ == "__main__":
    subsets = ["FN", "FP", "TP", "TN"]
    for subset in subsets:
        if subset == "FN":
            ped_set = fn_occ_top_30
        elif subset == "FP":
            ped_set = fp_occ_top_30
        elif subset == "TP":
            ped_set = tp_occ_combined
        elif subset == "TN":
            ped_set = tn_occ_combined
        else:
            ped_set = []

        for ped in tqdm(ped_set, desc=f"Processing {subset}"):
            extract_flow_frames_local(
                subset=subset,
                ped_id=ped,
                frames_to_show=32
            )

    print("Optical flow local extraction complete.")
```

```python
import os
import sys
import cv2
import pickle
import numpy as np
import pandas as pd
from tqdm import tqdm

from dataset.PIE_API import PIE
from tools.video_utils import VideoLoader
from tools.visualisation import draw_bbox
from tools.box_utils import jitter_bbox, squarify_bbox

# === CONFIGURATION ===
DATASET_DIR = r"..\PIE-Files"
CLIPS_DIR   = os.path.join(DATASET_DIR, "PIE_clips")
DB_PATH     = os.path.join(DATASET_DIR, "data_cache", "pie_database.pkl")

SAVE_DIR    = "../Categorical-Depth"
os.makedirs(SAVE_DIR, exist_ok=True)

SAVE_OUTPUT = False

ATTR_CSVS = {
    "FN": "../Data-Prep/annot-data/PIE_annot_attrb_val_fn.csv",
    "FP": "../Data-Prep/annot-data/PIE_annot_attrb_val_fp.csv",
    "TP": "../Data-Prep/annot-data/PIE_annot_attrb_val_tp.csv",
    "TN": "../Data-Prep/annot-data/PIE_annot_attrb_val_tn.csv"
}

# === DEPTH MODEL ===
sys.path.insert(0, os.path.join(os.getcwd(), "feature", "manydepth"))
from _interface import DepthEstimator

Depth = DepthEstimator(
    model_path=os.path.join(os.getcwd(), "feature", "manydepth", "CityScapes_MR"),
    intrinsics_json_path=os.path.join(os.getcwd(), "feature", "manydepth", "PIE_intrinsics.json")
)

# === PIE DATABASE ===
if not os.path.isfile(DB_PATH):
    print("Generating PIE database...")
    dataset = PIE(data_path=DATASET_DIR)
    dataset.generate_database()

data = pickle.load(open(DB_PATH, "rb"))


def extract_depth_frames_local(subset="FP", ped_id="3_2_295", frames_to_show=32):

    print(f"Processing depth local {subset} → {ped_id}")

    attr_df = pd.read_csv(ATTR_CSVS[subset])
    row     = attr_df[attr_df["id"] == ped_id]

    if len(row) == 0:
        print("Pedestrian not found in CSV")
        return

    critical_point = int(row["critical_point"].iloc[0])

    set_id, vid_id, ped_num = ped_id.split("_")
    set_name   = f"set0{set_id[-1]}" if not set_id.startswith("set") else set_id
    video_name = f"video_{int(vid_id):04d}"

    video_path = os.path.join(CLIPS_DIR, set_name, f"{video_name}.mp4")
    video      = VideoLoader(video_path)

    annots = data[set_name][video_name]["ped_annotations"].get(ped_id)
    frames = annots["frames"]

    crit_idx   = frames.index(critical_point)
    start_idx  = max(0, crit_idx - (frames_to_show - 1))
    seq_frames = frames[start_idx:crit_idx + 1]

    ped_outdir = os.path.join(SAVE_DIR, f"{subset}/{ped_id}_depth_local")
    if SAVE_OUTPUT:
        os.makedirs(ped_outdir, exist_ok=True)

    TARGET_DIM    = (224, 224)
    sampling_time = -1
    source_image  = None

    for frame_idx in seq_frames:
        frame = video[frame_idx]
        t     = annots["frames"].index(frame_idx)
        bbox  = list(map(int, annots["bbox"][t][0:4]))

        # === Source image cadence — mirrors depth extraction script ===
        sampling_time += 1
        if (sampling_time == 0) or (not sampling_time % 7):
            source_image = frame.copy()

        # === Run depth model on full frame ===
        depth_list    = Depth.run(source_image, frame)
        depth_heatmap = cv2.cvtColor(depth_list[1], cv2.COLOR_BGR2RGB)
        depth_heatmap = cv2.resize(depth_heatmap, (frame.shape[1], frame.shape[0]))

        # === Overlay depth heatmap on original frame ===
        overlay = cv2.addWeighted(frame, 0.1, depth_heatmap, 0.9, 0)

        frame_h, frame_w = frame.shape[:2]
        x1, y1, x2, y2  = bbox

        # === Local context crop — jitter + squarify matching model input pipeline ===
        jitter_box = jitter_bbox(video.size, [bbox.copy()], mode="enlarge", ratio=1.0)[0]
        local_box  = squarify_bbox(jitter_box, 1.0, frame_w)
        local_box  = tuple(map(int, local_box[0:4]))

        crop_region = overlay[
            local_box[1]:local_box[3],
            local_box[0]:local_box[2]
        ]
        
        # Resize cropped region to 224x224
        crop_region_resized = cv2.resize(
            crop_region,
            TARGET_DIM,
            interpolation=cv2.INTER_LINEAR
        )

        # === Remap bbox into local crop coordinate space ===
        scale_x = TARGET_DIM[0] / (local_box[2] - local_box[0])
        scale_y = TARGET_DIM[1] / (local_box[3] - local_box[1])

        mapped_bbox = [
            max(0,              int((x1 - local_box[0]) * scale_x)),
            max(0,              int((y1 - local_box[1]) * scale_y)),
            min(TARGET_DIM[0],  int((x2 - local_box[0]) * scale_x)),
            min(TARGET_DIM[1],  int((y2 - local_box[1]) * scale_y))
        ]

        annotated = draw_bbox(
            crop_region_resized,
            mapped_bbox,
            caption=ped_id,
            thickness=2,
            color=(255, 0, 0)
        )

        cv2.imshow("Depth Local + BBox", annotated)

        if SAVE_OUTPUT:
            out_path = os.path.join(
                ped_outdir,
                f"frame_{frame_idx}_depth_bbox_local.jpg"
            )
            cv2.imwrite(out_path, annotated)

        key = cv2.waitKey(1)
        if key == ord("q"):
            break

    cv2.destroyAllWindows()
    print(f"Finished depth local {subset}/{ped_id}")


if __name__ == "__main__":
    subsets = ["FN", "FP", "TP", "TN"]
    for subset in subsets:
        if subset == "FN":
            ped_set = fn_occ_top_30
        elif subset == "FP":
            ped_set = fp_occ_top_30
        elif subset == "TP":
            ped_set = tp_occ_combined
        elif subset == "TN":
            ped_set = tn_occ_combined
        else:
            ped_set = []

        for ped in tqdm(ped_set, desc=f"Processing {subset}"):
            extract_depth_frames_local(
                subset=subset,
                ped_id=ped,
                frames_to_show=32
            )

    print("Depth local extraction complete.")
```

#### Semantic Segmentation (Global)

```python
import os
import cv2
import pickle
import numpy as np
import pandas as pd
from tqdm import tqdm

from dataset.PIE_API import PIE
from tools.video_utils import VideoLoader
from tools.visualisation import draw_bbox
from feature.DeepLabV3Plus._interface import SemanticEstimator


# CONFIGURATION
DATASET_DIR = r"..\PIE-Files"
CLIPS_DIR = os.path.join(DATASET_DIR, "PIE_clips")
DB_PATH = os.path.join(
    DATASET_DIR,
    "data_cache",
    "pie_database.pkl"
)

SAVE_DIR = "../Full-Scale-Semantic"
os.makedirs(SAVE_DIR, exist_ok=True)

SAVE_OUTPUT = False

ATTR_CSVS = {
    "FN": "../Data-Prep/annot-data/PIE_annot_attrb_val_fn.csv",
    "FP": "../Data-Prep/annot-data/PIE_annot_attrb_val_fp.csv",
    "TP": "../Data-Prep/annot-data/PIE_annot_attrb_val_tp.csv",
    "TN": "../Data-Prep/annot-data/PIE_annot_attrb_val_tn.csv"
}

# LOAD SEMANTIC SEGMENTATION MODEL
CHECKPOINT = os.path.join(
    os.getcwd(),
    "feature",
    "DeepLabV3Plus",
    "checkpoints",
    "best_deeplabv3plus_resnet101_cityscapes_os16.pth"
)

Semantic = SemanticEstimator(ckpt=CHECKPOINT)

# LOAD PIE DATABASE
if not os.path.isfile(DB_PATH):
    print("Generating PIE database...")
    dataset = PIE(data_path=DATASET_DIR)
    dataset.generate_database()

data = pickle.load(open(DB_PATH, "rb"))

# FULL-SCALE SEMANTIC EXTRACTION FUNCTION
def extract_semantic_frames_fullscale(
    subset="FP",
    ped_id="3_2_295",
    frames_to_show=32
):
    print(f"Processing semantic full-scale {subset} → {ped_id}")

    # Load attribute metadata
    attr_df = pd.read_csv(ATTR_CSVS[subset])
    row = attr_df[attr_df["id"] == ped_id]
    if len(row) == 0:
        print("Pedestrian not found in CSV")
        return

    # Determine temporal window relative to prediction moment
    critical_point = int(row["critical_point"].iloc[0])
    set_id, vid_id, ped_num = ped_id.split("_")
    set_name = f"set0{set_id[-1]}" if not set_id.startswith("set") else set_id
    video_name = f"video_{int(vid_id):04d}"
    video_path = os.path.join(
        CLIPS_DIR,
        set_name,
        f"{video_name}.mp4"
    )

    video = VideoLoader(video_path)
    annots = data[set_name][video_name]["ped_annotations"].get(ped_id)
    frames = annots["frames"]
    crit_idx = frames.index(critical_point)
    start_idx = max(0, crit_idx - (frames_to_show - 1))
    seq_frames = frames[start_idx:crit_idx + 1]

    # Match bbox pipeline directory structure
    ped_outdir = os.path.join(
        SAVE_DIR,
        subset,
        f"{ped_id}_full"
    )

    if SAVE_OUTPUT:
        os.makedirs(ped_outdir, exist_ok=True)

    # Target output resolution (same as bbox full-scale)
    TARGET_W = 1280
    TARGET_H = 720

    for frame_idx in seq_frames:
        frame = video[frame_idx]
        t = annots["frames"].index(frame_idx)
        bbox = list(map(int, annots["bbox"][t][0:4]))
        frame_h, frame_w = frame.shape[:2]

        # Aspect ratio preserving resize
        scale = min(
            TARGET_W / frame_w,
            TARGET_H / frame_h
        )

        out_w = int(frame_w * scale)
        out_h = int(frame_h * scale)

        frame_resized = cv2.resize(
            frame,
            (out_w, out_h),
            interpolation=cv2.INTER_AREA
        )

        # Run semantic segmentation on resized frame
        semantic_map = Semantic.run(frame_resized)
        semantic_map = cv2.resize(
            semantic_map,
            (out_w, out_h),
            interpolation=cv2.INTER_NEAREST
        )

        # Overlay segmentation on RGB image
        overlay = cv2.addWeighted(
            frame_resized,
            0.45,
            semantic_map,
            0.55,
            0
        )

        # Scale bounding box coordinates
        scaled_bbox = [
            int(bbox[0] * scale),
            int(bbox[1] * scale),
            int(bbox[2] * scale),
            int(bbox[3] * scale)
        ]

        annotated = draw_bbox(
            overlay,
            scaled_bbox,
            caption=ped_id,
            thickness=2,
            color=(255, 0, 0)
        )

        cv2.imshow(
            "Semantic Full Scene + BBox",
            annotated
        )

        # Save output with consistent filename pattern
        if SAVE_OUTPUT:
            out_path = os.path.join(
                ped_outdir,
                f"frame_{frame_idx}_sem_bbox_full.jpg"
            )

            cv2.imwrite(
                out_path,
                annotated
            )


        key = cv2.waitKey(1)
        if key == ord("q"):
            break

    cv2.destroyAllWindows()
    print(f"Finished semantic full-scale {subset}/{ped_id}")

# BATCH EXTRACTION LOOP
if __name__ == "__main__":
    subsets = ["FN", "FP", "TP", "TN"]
    for subset in subsets:
        if subset == "FN": ped_set = fn_occ_top_30
        elif subset == "FP": ped_set = fp_occ_top_30
        elif subset == "TP": ped_set = tp_occ_combined
        elif subset == "TN": ped_set = tn_occ_combined
        else:  ped_set = []

        for ped in tqdm(ped_set,desc=f"Processing {subset}"):
            extract_semantic_frames_fullscale(
                subset=subset,
                ped_id=ped,
                frames_to_show=32
            )

    print("Semantic full-scale extraction complete.")
```
